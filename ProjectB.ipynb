```python
import numpy as np
```

*Inizialization of parameters*

```python
C = np.array ([[4, -1], [-1, 4]])
mean = [0, 0]

a = [1, 3, 10]
delta = 3
n = 5000

Y = np.zeros (n)
L = np.zeros (3)
U = np.zeros (3)
```

*Generating n standard normal random variables*

```python
seed = 42
gen = np.random.default_rng(seed)
Z = gen.multivariate_normal (mean, C, size=n)
```

### (a)

```python
for i in range (3):
    # generate Y = 1 {Z that belongs to A}
    Y = np.logical_and (Z[:, 0]>a[i], Z[:, 0]>a[i])

    # calculate the mean estimator of Y
    mu = np.mean(Y)

    # calculate the standard deviation estimator of Y
    X = (Y - mu)**2
    sd = np.sqrt ( sum (X) / (n-1) )

    # calculate the Lower bound of the 95% confidence interval
    L[i] = mu - 1.95 * sd / np.sqrt(n)

    # calculate the Lower bound of the 95% confidence interval
    U[i] = mu + 1.95 * sd / np.sqrt(n) 
```

*Print the 95% confidence interval for each a*

```python
for i in range(3):
    print (f'The 95% confidence interval with a = {a[i]} is [{L[i]}, {U[i]}]')
```
